## Sub-step 1.3b: Visualize Nulls (Bar Chart)

## 1.3c: Resolve Null Values
|Resolve null values by appropriate strategy based on the datatype & percentage of null values|

```python
null_percent = (df.isnull().sum()/len(df))*100
null_percent = null_percent[null_percent > 0]
print("Percentage of nulls per column:")
print(null_percent)
```

```python
## Drop negligible-null rows (vin, mmr, sellingprice, saledate, odometer)

df = df.dropna(subset=['vin', 'mmr', 'sellingprice', 'saledate', 'odometer'])
print("Shape after dropping negligible nulls:", df.shape)
```

```python
## Fill categorical columns with Mode (color, interior, make, model, body)

for col in ['color', 'interior', 'make', 'model', 'body']:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Nulls remaining in color, interior, make, model, body:")
print(df[['color', 'interior', 'make', 'model', 'body']].isnull().sum())
```

```python
##Fill categorical columns with "Unknown" (trim, transmission)

for col in ['trim', 'transmission']:
    df[col] = df[col].fillna('Unknown')

print("Nulls remaining in trim, transmission:")
print(df[['trim', 'transmission']].isnull().sum())
```

```python
## Fill Numeric Column (condition) with Median

df['condition'] = df['condition'].fillna(df['condition'].median())
print("Nulls remaining in condition:", df['condition'].isnull().sum())
```

```python
## Confirm No Nulls Remain Anywhere

print("Total nulls remaining in entire dataset:")
print(df.isnull().sum().sum())
```

## 3.2 Avg selling price by year

```python
avg_price_by_year = df.groupby('year')['sellingprice'].mean()

plt.figure(figsize=(8, 4))
avg_price_by_year.plot(kind='bar')
plt.title("Average Selling Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Selling Price")
plt.show()
```

**Insight:** Plot choice: A bar chart was used rather than a scatter plot, since year is a discrete, limited set of categories (not a continuous variable) — a bar chart makes it easy to compare the single average price per year, whereas a scatter plot would be more appropriate for showing the relationship between two continuous variables. Average selling price shows a clear upward trend as cars get newer, consistent with the positive correlation (0.59) found between year and sellingprice in Task 3.1. Older cars (1990s) average under  USD 2,000, while 2015 models average around $26,000. The spike at 1982 is likely a small-sample anomaly (very few cars from that year), not a genuine trend, since price rises smoothly and consistently from the 1990s onward.

##  3.7: Box Plot of Selling Price by Color

```python
import seaborn as sns

plt.figure(figsize=(14, 6))
sns.boxplot(x='color', y='sellingprice', data=df)
plt.title("Distribution of Selling Price by Color")
plt.xlabel("Color")
plt.ylabel("Selling Price")
plt.xticks(rotation=45)
plt.show()
```

```python
valid_colors = ['white', 'gray', 'black', 'red', 'silver', 'blue', 'brown',
                 'beige', 'purple', 'burgundy', 'gold', 'yellow', 'green',
                 'charcoal', 'orange', 'off-white', 'turquoise', 'pink', 'lime']

df['color'] = df['color'].where(df['color'].isin(valid_colors), np.nan)
df['color'] = df['color'].fillna(df['color'].mode()[0])

print("Number of unique values in color:", df['color'].nunique())
```

-- ### Insight (before outlier removal)
Most colors show similar median prices ($10,000-$17,000), but every color has a long tail of outliers reaching $150,000+. The green category includes an extreme outlier near USD 230,000, consistent with the anomalous 2014 Ford Escape found in Task 2.4. This suggests outliers should be removed for a clearer view of typical price distribution.

## 3.7 (continued): Box Plot After Outlier Removal

```python
Q1 = df['sellingprice'].quantile(0.25)
Q3 = df['sellingprice'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df_no_outliers = df[(df['sellingprice'] >= lower_limit) & (df['sellingprice'] <= upper_limit)]

print("Rows before removing outliers:", len(df))
print("Rows after removing outliers:", len(df_no_outliers))

plt.figure(figsize=(14, 6))
sns.boxplot(x='color', y='sellingprice', data=df_no_outliers)
plt.title("Distribution of Selling Price by Color (Outliers Removed)")
plt.xlabel("Color")
plt.ylabel("Selling Price")
plt.xticks(rotation=45)
plt.show()
```

-- ## Insight (after outlier removal): Removing outliers via the IQR method dropped ~16,350 rows (~2.9% of the dataset) — the extreme high-value listings like the ~$230,000 green car identified earlier. With those gone, the y-axis compresses to a realistic price range, making it much easier to compare typical price distributions across colors. Median prices stay broadly similar across colors even after cleanup, confirming the original color-based pricing pattern was genuine and not driven by the outliers — the outliers were mainly distorting the visual scale, not the underlying typical price relationships.
